**[Task 2.2.]()** Implement the two functions `sobel_h` and `sobel_v` for horizontal and vertical $3 \times 3$ Sobel filtering, respectively:

```python
def sobel_h(img_in):
    img_in = img_in.astype(float)
    img_out = img_in.copy()
    dis = 1
    for p in np.ndindex(img_in.shape): #für alle einzelnen Pixel -> Rand beachten!
        if p[0] >= dis and p[0] < (img_in.shape[0]-dis) and p[1] >= dis and p[1] < (img_in.shape[1]-dis): #für alle Zellen die NICHT am Rand sind - Rand ist jeweils size/2-0,5
            listp = list(p)
            row = listp[0]
            column = listp[1]
            value = (1/8)*(img_in[row-1, column-1]-img_in[row-1, column+1]+2*img_in[row, column-1]-2*img_in[row, column+1]+img_in[row+1, column-1]-img_in[row+1, column+1])
            img_out[p] = value
    return img_out
    
def sobel_v(img_in):
    img_in = img_in.astype(float)
    img_out = img_in.copy()
    dis = 1
    for p in np.ndindex(img_in.shape): #für alle einzelnen Pixel -> Rand beachten!
        if p[0] >= dis and p[0] < (img_in.shape[0]-dis) and p[1] >= dis and p[1] < (img_in.shape[1]-dis): #für alle Zellen die NICHT am Rand sind - Rand ist jeweils size/2-0,5
            listp = list(p)
            row = listp[0]
            column = listp[1]
            value = (1/8)*(img_in[row-1, column-1]+2*img_in[row-1, column]+img_in[row-1, column+1]-(img_in[row+1, column-1]+2*img_in[row+1, column]+img_in[row+1, column+1]))
            img_out[p] = value
    return img_out
```

```python
def sobel(image):
    image = np.asarray(image, dtype=float)

    # Horizontale Ableitung
    Gx = (
        -image[:-2, :-2] + image[:-2, 2:]
        - 2*image[1:-1, :-2] + 2*image[1:-1, 2:]
        - image[2:, :-2] + image[2:, 2:]
    )

    # Vertikale Ableitung
    Gy = (
        -image[:-2, :-2] - 2*image[:-2, 1:-1] - image[:-2, 2:]
        + image[2:, :-2] + 2*image[2:, 1:-1] + image[2:, 2:]
    )

    # Gradient berechnen
    G = np.sqrt(Gx**2 + Gy**2)

    return G

```
