**[Task 2.1.]()** Implement Otsu thresholding as a re-usable function:

```python
def otsu_thresholding(imgf):
    img8 = (imgf * 255).round().astype(np.uint8)
    h = np.zeros(256)
    for i in range (0,255):
        j = np.array(img8 == i).sum()
        h[i]=j
    list_T = []

    for t in range (1,255):
        T = h[:t].sum()*np.var(h[:t])+h[t:].sum()*np.var(h[t:])
        list_T.append(T)
    threshold = list_T.index(min(list_T))+1
    
    seg = np.array(img8 > threshold)
    
    return seg
```

**[Task 2.2.]()** Implement a re-usable function to compute the Dice coefficient:

```python
def compute_dice(bin1, bin2):
    bin1 = np.asarray(bin1>0, dtype = bool)
    bin2 = np.asarray(bin2, dtype = bool)
    assert bin1.dtype == bool
    assert bin2.dtype == bool
    
    GH = np.logical_and(bin1, bin2).sum()
    dice = 2*GH/(bin1.sum()+bin2.sum())
    
    return dice
```
