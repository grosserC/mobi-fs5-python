**[Task 1.2 (d).]()** Binarize the image by applying an intensity threshold:

```python
plt.hist(img)

T=0.10
img_out = np.array(img > T)

plt.figure()
plt.imshow(img_out)
plt.colorbar()

```

**[Task 1.2 (e).]()** Load the ground truth:

```python
groundtruth = plt.imread('data/NIH3T3/gt/0.png')

plt.figure()
plt.imshow(groundtruth, cmap = "gray")
plt.colorbar()
```

```python
H = np.asarray(groundtruth>0, dtype = bool)
G = np.asarray(img_out, dtype = bool)
GH = np.logical_and(G,H).sum()
dice = 2*GH/(G.sum()+H.sum())
print(dice)
```

Compute and print the Dice coefficient:

```python
G = groundtruth.astype(np.uint8)
H = img_out.astype(np.uint8)

dice = (2*(G*H).sum())/(G.sum() + H.sum())
print(dice)
```
